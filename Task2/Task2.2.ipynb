```python
import numpy as np
import pandas as pd

'''Done using ridge regression as it automatically adds a penalty term to the loss function.This discourages large coefficients, making the model more stable and less sensitive to noise or multicollinearity.'''
train = pd.read_csv('fishing_logs.csv')
test = pd.read_csv('prediction_logs.csv')

X_train= train[['Depth Pressure (kPa)','Water Temperature (°C)','Catch Density Metric']].values
y_train= train['Catch Quality Rating'].values
X_test= test[['Depth Pressure (kPa)','Water Temperature (°C)','Catch Density Metric']].values


#Add bias column (for intercept)
X_train = np.c_[np.ones(X_train.shape[0]), X_train]
X_test = np.c_[np.ones(X_test.shape[0]), X_test]

#Compute weights using matrix method
# w = (X^T X+lamda*I)^(-1) X^T y
lambda_reg=10000#for preventing overfitting using ridge regreszion
I=np.eye(X_test.shape[1])#Identity matrix
I[0,0]=0
w = np.linalg.inv(X_train.T.dot(X_train)+lambda_reg*I).dot(X_train.T).dot(y_train)


y_pred_train = X_train.dot(w)
y_pred_test=X_test.dot(w)
pd.set_option("display.max_rows", None)  # show all rows
print(pd.Series(y_pred_test))

# Load the existing Excel file
file_path = "prediction_logs_results.xlsx"
df = pd.read_excel(file_path)

# Adding predictions as a new column
df["Catch quality ratings"] = y_pred_test

# Save back to the same file (overwrite)
df.to_excel(file_path, index=False)

df.to_csv("prediction_logs_results.csv", index=False)

print(df.head())

```
