```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from LinearRegression import LinearRegression

regr=LinearRegression()
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')

train = train.dropna()
test = test.dropna()

X_train = train[[ 'low', 'beta_indicator','high','open',
                 'market_sentiment','liquidity_ratio','oscillator_value'
                 ]].values
y_train = train['close'].values
X_test = test[[
              'low', 'beta_indicator','high','open',
                 'market_sentiment','liquidity_ratio','oscillator_value']].values
y_test = test['close'].values

#Normalization
mean_X = X_train.mean(axis=0)
mean_Y = y_train.mean(axis=0)
sd_X = X_train.std(axis=0)
sd_Y = y_train.std(axis=0)

l, = sd_X.shape
for i in range (l):
    if(sd_X[i].round() == 0):
        print(i)

sd_X[sd_X == 0] = 1
if sd_Y == 0:
    sd_Y = 1.0

X_train = (X_train - mean_X)/sd_X
y_train = (y_train - mean_Y)/sd_Y
#Normalization of testing
mean_X = X_test.mean(axis=0)
mean_Y = y_test.mean(axis=0)
sd_X = X_test.std(axis=0)
sd_Y = y_test.std(axis=0)

l, = sd_X.shape
for i in range (l):
    if(sd_X[i].round() == 0):
        print(i)

sd_X[sd_X == 0] = 1
if sd_Y == 0:
    sd_Y = 1.0

X_test = (X_test - mean_X)/sd_X
y_test = (y_test - mean_Y)/sd_Y
#calculating weights
regr.fit(X_train, y_train)
#calculating predictions
y_pred_test=regr.predict(X_test)



#Compute accuracy metrics manually
ss_res = np.sum((y_test - y_pred_test) ** 2)
ss_tot = np.sum((y_test - np.mean(y_test)) ** 2)
r2_score = 1 - (ss_res / ss_tot)

mae = np.mean(np.abs(y_test - y_pred_test))
rmse = np.sqrt(np.mean((y_test - y_pred_test) ** 2))

print(f"R² Score on Testing Data: {r2_score:.4f}")
print(f"Mean Absolute Error: {mae:.4f}")
print(f"Root Mean Squared Error: {rmse:.4f}")

print(y_pred_test)


```
